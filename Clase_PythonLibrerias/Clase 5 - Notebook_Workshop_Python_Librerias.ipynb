Hasta ahora son todas variables / funciones sueltas \(que NO están mal, pero depende que objetivo / finalidad tenga\) pero ahora vamos a plantear una lógica de negocio, o en este caso un juego \(simil pokémon\)\.

```
# creamos una clase que se llame "Pokemon" y le damos características (nombre, vida, ataque)

class Pokemon:
    def __init__(self, nombre, vida, ataque):
        self.nombre = nombre
        self.vida = vida
        self.ataque = ataque
```

La Clase ya está creada, pero no hay ningún Pokémon "vivo", porque lo que hicimos fué crear la fábrica\(plantilla\) para crear pokemones

Ahora vamos a crear al primero Pokémon

Creamos un pokemon de nombre "Pikachu", con 100 de vida y 10 de ataque

```
Pokemon1 = Pokemon("Pikachu", 100, 10)
```

veamos los atributos \(características\) de nuestro primer pokemon

```
print(Pokemon1.nombre) #imprimimos el nombre del pokemon
print(Pokemon1.vida) #imprimimos la vida del pokemon
print(Pokemon1.ataque) #imprimimos el ataque del pokemon
```

Vamos a agregarle algunas acciones\(métodos\) al pokemon

```
class Pokemon:
    def __init__(self, nombre, vida, ataque):
        self.nombre = nombre
        self.vida = vida
        self.ataque = ataque

    def atacar(self):
        print("estoy atacando")
```

Llevamos a cabo la acción\(método\)

```
pokemon1 = Pokemon("Pikachu", 100, 10)
pokemon1.atacar()
```

# 2\. Limpieza y transformación de datos

En el material, esta etapa está asociada principalmente a pandas, polars y datatable\.
La idea es organizar, filtrar, renombrar, agrupar y preparar los datos para análisis posterior\.

### pandas

Utilidad: Manipulación y análisis de datos en tablas \(DataFrames\)\. Lectura de CSV/Excel/SQL, limpieza y transformación\.

Ejemplo: filtrado, creación de variables y agregación\.

**Referencia:** [McKinney (2010)](https://proceedings.scipy.org/articles/Majora-92bf1922-00a)  

```
try:
    base_df = cargar_encuesta()
    df_limpio = (
        base_df
          .rename(columns={"cumplio": "cumplimiento"})
          .assign(ahorro=lambda x: x["ingreso"] - x["gasto"])
    )

    resumen = df_limpio.groupby("grupo")[["ingreso", "gasto", "ahorro"]].mean().round(2)
    print(type(resumen))
    print(resumen)
except Exception as e:
    print("No se pudo ejecutar el ejemplo de limpieza.")
    print("Detalle:", e)
```

### polars  

Utilidad: Alternativa a pandas optimizada en Rust\. Muy rápida para análisis tabular en datasets grandes\.

Ejemplo: transformación tabular con sintaxis expresiva\.

**Referencia:** [Polars Docs](https://docs.pola.rs/)  

```
!pip install polars
```

```
try:
    import polars as pl

    pl_df = pl.read_csv("encuesta.csv")
    resultado_polars = (
        pl_df
        .with_columns((pl.col("ingreso") - pl.col("gasto")).alias("ahorro"))
        .group_by("grupo")
        .agg([
            pl.mean("ingreso").alias("prom_ingreso"),
            pl.mean("gasto").alias("prom_gasto"),
            pl.mean("ahorro").alias("prom_ahorro")
        ])
    )
    print(resultado_polars)
except Exception as e:
    print("Polars no está disponible o falló la ejecución.")
    print("Detalle:", e)
```
