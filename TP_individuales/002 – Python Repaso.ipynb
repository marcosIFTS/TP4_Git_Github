### Ejercicio 1 – Variables y operaciones

```
nombre='Marcos'
edad='35'
ciudad='Villa Luro'

print(f"Hola, mi nombre es {nombre}, tengo {edad} años y vivo en {ciudad}.")
```

```
numero1=35
numero2=7

suma= numero1 + numero2
resta= numero1 - numero2
multiplicacion= numero1 * numero2
division= numero1 / numero2

print(f"La suma de {numero1} y {numero2} es {suma}.")
print(f"La resta de {numero1} y {numero2} es {resta}.")
print(f"La multiplicación de {numero1} y {numero2} es {multiplicacion}.")
print(f"La división de {numero1} y {numero2} es {division}.")
```

### Ejercicio 2 – Condicionales

```
edad = int(input('Ingrese su edad: '))
if edad >= 18:
    print('Eres mayor de edad')
else:
    print('Eres menor de edad')
```

### Ejercicio 3 – Listas y bucles

```
lista = [10, 22, 8, 3, 15]

print("Los elementos de la lista son:", end=" ")

for i in lista:
    print(i , end=" ")

suma = sum(lista)

promedio = suma / len(lista)

print(f"\nLa suma de la lista es: {suma}")

print(f"El promedio de la lista es: {promedio}")
```
