### Ejercicio 2 – Dataset simple con Pandas

```

datos = {
    'estudiante': ['Ana', 'Juan', 'Carla','Pedro','Lucia'],
    'nota': [8, 6, 9, 5, 7],
}

examen = pd.DataFrame(datos)

promedio_nota = examen['nota'].mean()

maxima_nota = examen['nota'].max()

minima_nota = examen['nota'].min()

print(examen)
print(f"Promedio de notas: {promedio_nota}")
print(f"Mejor nota del examen: {maxima_nota}")
print(f"Peor nota del examen: {minima_nota}")
```

### Ejercicio 3 – Criterio de selección de herramientas

1\) pandas → porque permite leer lectura de archivos CSV\. Ideal para datasets pequeños y medianos\.

2\) Matplotlib → porque representa una base de gráficos 2D o gráficos complejos, como histogramas o graficos de densidad\.

3\) Scikit\-learn→ porque su API permite cambiar de un modelo a otro, tiene algoritmos de Machine Learning clásico: clasificación, regresión, clustering, reducción de dimensionalidad\.

4\) numpy → porque realiza operaciones sobre matrices como multiplicaciones matriciales o trasposiciones\.
